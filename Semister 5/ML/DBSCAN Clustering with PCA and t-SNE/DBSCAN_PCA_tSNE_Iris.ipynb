# DBSCAN Clustering with PCA and t-SNE
**Dataset:** Iris dataset (built into scikit-learn)

## Aim
To apply DBSCAN clustering to a dataset and visualize the resulting clusters using PCA and t-SNE.

## Theory
DBSCAN (Density-Based Spatial Clustering of Applications with Noise) groups points that are closely packed together. It uses:
- **eps:** maximum distance between two samples for them to be considered neighbours.
- **min_samples:** minimum number of points required to form a dense region.
- **Noise (`-1`):** points that do not belong to any cluster.

PCA and t-SNE are dimensionality-reduction techniques used here to display the four-dimensional Iris data in two dimensions. DBSCAN is fitted on the scaled original features, not on the 2D plots.

## 1. Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score
```

## 2. Load the dataset

```python
# Load the Iris dataset
iris = load_iris()
X = iris.data
feature_names = iris.feature_names

df = pd.DataFrame(X, columns=feature_names)
df.head()
```

## 3. Standardize the features

```python
# Scaling is important because DBSCAN uses distance calculations.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 4. Apply DBSCAN

```python
# eps and min_samples are DBSCAN parameters.
dbscan = DBSCAN(eps=0.8, min_samples=5)
labels = dbscan.fit_predict(X_scaled)

n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = np.sum(labels == -1)

print("Number of clusters:", n_clusters)
print("Number of noise points:", n_noise)
print("Cluster labels:", np.unique(labels))
```

## 5. Visualize DBSCAN results using PCA

```python
# Reduce the scaled data from 4 dimensions to 2 dimensions.
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Variance retained by 2 PCA components:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

plt.figure(figsize=(8, 6))
for cluster in sorted(set(labels)):
    mask = labels == cluster
    label_name = "Noise" if cluster == -1 else f"Cluster {cluster}"
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1],
                s=50, label=label_name)
plt.title("DBSCAN Clusters - PCA (Iris Dataset)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 6. Visualize DBSCAN results using t-SNE

```python
# t-SNE maps the data into 2 dimensions for visualization.
tsne = TSNE(n_components=2, random_state=42,
            perplexity=30, init="pca", learning_rate="auto")
X_tsne = tsne.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
for cluster in sorted(set(labels)):
    mask = labels == cluster
    label_name = "Noise" if cluster == -1 else f"Cluster {cluster}"
    plt.scatter(X_tsne[mask, 0], X_tsne[mask, 1],
                s=50, label=label_name)
plt.title("DBSCAN Clusters - t-SNE (Iris Dataset)")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 7. Result and conclusion
DBSCAN was applied to the standardized Iris dataset. The algorithm assigned each sample to a density-based cluster or marked it as noise (`-1`). PCA and t-SNE were then used to visualize the DBSCAN labels in two dimensions.

**Note:** The cluster shapes in PCA and t-SNE are visual projections. DBSCAN performed clustering on the original standardized four-feature data. The number of clusters and noise points depends on the selected `eps` and `min_samples` values.
