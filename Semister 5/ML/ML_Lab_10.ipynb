```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
centres = np.array ([
    [25, 25], [25, 25],
    [75, 25], [75, 75]
])

X, _ = make_blobs(
    n_samples = 300,
    centers = centres,
    cluster_std = 8,
    random_state = 42
)

X[:, 0] = np.clip(X[:, 0], 1, 100)
X[:, 1] = np.clip(X[:, 1], 1, 100)

df = pd.DataFrame (
    X, columns = ["AnnualIncome", "SpendingScore"]
)

df.to_csv("customer_segmentation.csv", index = False)

print("Dataset Shape: ", df.shape)
print(df.head())
print(df.describe(). round(2))
```

```python
plt.figure(figsize = (7, 5))
plt.scatter(df["AnnualIncome"], df["SpendingScore"], alpha = 0.75)
plt.xlabel("Annual Income (simulated units)")
plt.ylabel("Spending Score (simulated units)")
plt.title("Customer Data Before Clustering")
plt.grid(alpha = 0.25)
plt.tight_layout()
plt.show()
```
