```python
# Required Python libraries are imported for data manipulation and analysis
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
# The Iris dataset is loaded using the Scikit-Learn
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
# The different components available in the Iris dataset are displayed
print("Feature Names: ")
print(iris.feature_names)

print("\nTarget Classes: ")
print(iris.target_names)
```

```python
# The feature data is converted into Pandas DataFrame for further analysis
df = pd.DataFrame(
    iris.data,
    columns = iris.feature_names
)
```
